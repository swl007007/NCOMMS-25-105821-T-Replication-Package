# Figure S1: Supplementary Figure - Cross-Validation Methodology

## Purpose
Create **supplementary figure** documenting the cross-validation strategy and fold assignment methodology.

## Methodology
1. Visualize temporal vs. random fold splitting
2. Show expanding window CV approach
3. Illustrate train/test partitioning across folds
4. Document fold assignment for reproducibility

## Expected Outputs
- Diagram showing CV methodology
- Visual comparison of different validation strategies
- Supplementary material for methods section

## Note
This figure supports the manuscript's methodology transparency and reproducibility requirements.

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# Define colors (approximating from the image)
color_train = '#D6EAF8'  # Light blue
color_val = '#D5F5E3'    # Light green
color_hyper = '#FEF9E7'  # Light yellow
color_holdout = '#EAEDED' # Light grey
color_arrow = '#F5B7B1'  # Light red
text_color = '#2C3E50' # Dark grey/black for text

# Create the figure and axes
fig, ax = plt.subplots(figsize=(12, 8))
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.axis('off') # Turn off the axis lines and labels

# --- Plotting the expanding training/validation sets ---
y_start = 8.5
bar_height = 0.8
gap = 0.3 # Gap between bars

# Set 1: Train (2017), Val (2018)
ax.add_patch(patches.Rectangle((1, y_start), 2, bar_height, facecolor=color_train, edgecolor='grey'))
ax.text(2, y_start + bar_height/2, "Traning Set\n(2017)", ha='center', va='center', fontsize=9, color=text_color)
ax.add_patch(patches.Rectangle((3, y_start), 1, bar_height, facecolor=color_val, edgecolor='grey'))
ax.text(3.5, y_start + bar_height/2, "Validation\n(2018)", ha='center', va='center', fontsize=9, color=text_color)

# Set 2: Train (2017-2018), Val (2019)
y_start -= (bar_height + gap)
ax.add_patch(patches.Rectangle((1, y_start), 3, bar_height, facecolor=color_train, edgecolor='grey'))
ax.text(2.5, y_start + bar_height/2, "Traning Set\n(2017-2018)", ha='center', va='center', fontsize=9, color=text_color)
ax.add_patch(patches.Rectangle((4, y_start), 1, bar_height, facecolor=color_val, edgecolor='grey'))
ax.text(4.5, y_start + bar_height/2, "Validation\n(2019)", ha='center', va='center', fontsize=9, color=text_color)

# Set 3: Train (2017-2019), Val (2020)
y_start -= (bar_height + gap)
ax.add_patch(patches.Rectangle((1, y_start), 4, bar_height, facecolor=color_train, edgecolor='grey'))
ax.text(3, y_start + bar_height/2, "Traning Set\n(2017-2019)", ha='center', va='center', fontsize=9, color=text_color)
ax.add_patch(patches.Rectangle((5, y_start), 1, bar_height, facecolor=color_val, edgecolor='grey'))
ax.text(5.5, y_start + bar_height/2, "Validation\n(2020)", ha='center', va='center', fontsize=9, color=text_color)

# Set 4: Train (2017-2020), Val (2021)
y_start -= (bar_height + gap)
ax.add_patch(patches.Rectangle((1, y_start), 5, bar_height, facecolor=color_train, edgecolor='grey'))
ax.text(3.5, y_start + bar_height/2, "Traning Set\n(2017-2020)", ha='center', va='center', fontsize=9, color=text_color)
ax.add_patch(patches.Rectangle((6, y_start), 1, bar_height, facecolor=color_val, edgecolor='grey'))
ax.text(6.5, y_start + bar_height/2, "Validation\n(2021)", ha='center', va='center', fontsize=9, color=text_color)

# --- Plotting the Hyperparameter Tuning box and arrow ---
hyper_y_start = y_start + (8.5 - y_start + bar_height) / 2 - 0.5 # Center vertically with the block of sets
hyper_box_x = 7.5
hyper_box_width = 2
hyper_box_height = 1
ax.add_patch(patches.Rectangle((hyper_box_x, hyper_y_start), hyper_box_width, hyper_box_height, facecolor=color_hyper, edgecolor='grey'))
ax.text(hyper_box_x + hyper_box_width/2, hyper_y_start + hyper_box_height/2, "Hyperparameter Tuning", ha='center', va='center', fontsize=10, color=text_color)

# Arrow from Hyperparameter Tuning
arrow_start_y = hyper_y_start -0.1
arrow_end_y = y_start - (bar_height + gap) + 0.1 # Pointing towards the next stage
arrow_x = hyper_box_x + hyper_box_width/2
ax.arrow(arrow_x, arrow_start_y, 0, arrow_end_y - arrow_start_y,
         head_width=0.1, head_length=0.2, fc=color_arrow, ec=color_arrow, length_includes_head=True, lw=1.5)

# Arrow pointiung Hyperparameter Tuning
arrow_start_hyper_y = hyper_y_start + 0.5
arrow_start_hyper_x = hyper_box_x - 0.2

ax.arrow(arrow_start_hyper_x, arrow_start_hyper_y, 0.16, 0,
         color='black',lw=1.5)

# --- Plotting the final Training and Hold-out Set ---
final_y_start = y_start - (bar_height + gap) - (bar_height + gap) # Position below the arrow
final_bar_height = 1.2
train_hold_gap = 0.05

ax.add_patch(patches.Rectangle((1, final_y_start- 0.2), 6, final_bar_height, facecolor=color_train, edgecolor='grey'))
ax.text(1 + 6/2, final_y_start + final_bar_height/2-0.1, "Traning Set (2017-2021)", ha='center', va='center', fontsize=10, color=text_color)
ax.add_patch(patches.Rectangle((1 + 6 + train_hold_gap, final_y_start- 0.2), 2, final_bar_height, facecolor=color_holdout, edgecolor='grey'))
ax.text(0.5 + 6 + train_hold_gap + 3/2, final_y_start + final_bar_height/2-0.1, "Hold-out Set (2022)", ha='center', va='center', fontsize=10, color=text_color)

# --- Brackets ---
# Vertical bracket for the expanding sets
bracket_x = hyper_box_x-0.2
bracket_top_y = 8.5 + bar_height + 0.1
bracket_bottom_y = y_start - 0.1
ax.plot([bracket_x, bracket_x], [bracket_bottom_y, bracket_top_y], color='black', lw=1.5)
ax.plot([bracket_x-0.1, bracket_x], [bracket_top_y, bracket_top_y], color='black', lw=1.5) # Top tick
ax.plot([bracket_x-0.1, bracket_x], [bracket_bottom_y, bracket_bottom_y], color='black', lw=1.5) # Bottom tick

# Horizontal bracket for the final sets
h_bracket_y = final_y_start - 0.4
h_bracket_start_x = 0.5 - 0.1
h_bracket_end_x = 0.5 + 6 + train_hold_gap + 3 + 0.1
ax.plot([h_bracket_start_x, h_bracket_end_x], [h_bracket_y, h_bracket_y], color='black', lw=1.5)
ax.plot([h_bracket_start_x, h_bracket_start_x], [h_bracket_y, h_bracket_y+0.1], color='black', lw=1.5) # Left tick
ax.plot([h_bracket_end_x, h_bracket_end_x], [h_bracket_y, h_bracket_y+0.1], color='black', lw=1.5) # Right tick

plt.savefig(
    'produced_graph/PictureS2.jpg', 
    dpi=300,
    format='jpeg',
    bbox_inches='tight',
    pil_kwargs={'compression': 'lzw'}
)

plt.show()
```
