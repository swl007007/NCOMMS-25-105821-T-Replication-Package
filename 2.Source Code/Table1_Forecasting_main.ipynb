# Table 1: Forecasting Analysis - Main Results

## Purpose
This notebook trains and evaluates XGBoost regression models for **forecasting food crisis phases 12 months ahead** using historical data only (no concurrent real-time variables).

## Methodology
1. **Data Loading**: Load preprocessed forecasting dataset with 12-month lagged features
2. **Temporal Split**: Train on data before 2022-01-01, test on data from 2022-01-01 onward
3. **Multi-Phase Modeling**: Train 4 separate XGBoost regressors for cumulative phase thresholds:
   - Phase 2+ (phase2_worse)
   - Phase 3+ (phase3_worse) 
   - Phase 4+ (phase4_worse)
   - Phase 5 (phase5_worse)
4. **Phase Classification**: Convert probability predictions to overall IPC phases using 20% threshold rule
5. **Model Evaluation**: Calculate accuracy, sensitivity, precision, and R² for crisis phases (3+)
6. **Interpretability**: Extract SHAP values for feature importance analysis

## Expected Outputs
- Classification metrics (accuracy, sensitivity, precision, R²)
- SHAP feature importance values stored in `shape_values_df_ensemble`
